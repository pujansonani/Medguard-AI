# MEDGUARD AI: Exploratory Data Analysis (EDA)
## Multimodal ICU Deterioration Prediction Cohort

This notebook provides research-grade exploratory data analysis covering:
1. Cohort demographics and class balance
2. Physiological stream missingness profiles
3. Dynamic vital sign and biomarker trajectories (Survivors vs Non-Survivors)
4. Clinical text notes frequency and character distributions
5. Correlation structure across 24h summary features

```python
import os
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Add src to pythonpath
sys.path.insert(0, str(Path.cwd().parent / "src"))
from medguard.preprocessing.time_series import FEATURE_COLUMNS
```

```python
# Load Cohort and Timeseries
df_cohort = pd.read_csv("../data/processed/cohort.csv")
df_ts = pd.read_parquet("../data/synthetic/timeseries.parquet")
df_notes = pd.read_parquet("../data/synthetic/notes.parquet")

print(f"Cohort Size: {len(df_cohort):,} ICU Stays")
print(f"Mortality Rate (48h): {df_cohort['mortality_48h'].mean():.2%}")
df_cohort.head()
```

```python
# 1. Mortality Class Distribution
fig, ax = plt.subplots(figsize=(6, 4))
df_cohort['mortality_48h'].value_counts().plot(kind='bar', color=['#3ECFB2', '#FF5F56'], ax=ax)
ax.set_xticklabels(['Survived', 'Deteriorated'], rotation=0)
ax.set_title('48-Hour ICU Mortality Balance')
ax.set_ylabel('Patient Count')
plt.show()
```

```python
# 2. Missingness Analysis across 18 Hourly Streams
vital_cols = [c for c in FEATURE_COLUMNS if c in df_ts.columns]
miss = df_ts[vital_cols].isna().mean().sort_values()

plt.figure(figsize=(9, 6))
miss.plot(kind='barh', color='#0F2B46')
plt.title('Physiological Stream Missingness Rates (%)')
plt.xlabel('Fraction Missing')
plt.grid(True, axis='x', linestyle=':')
plt.show()
```
